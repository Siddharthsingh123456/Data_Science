# Module 26 Sales Forecasting Case Study

```

import pandas as pd, numpy as np, matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.arima.model import ARIMA
df = pd.read_csv('stores_sales_forecasting.csv', encoding='latin1')
df.head()
```

```

print(df.shape)
print(df.info())
print(df.describe())
print(df.isnull().sum())
```

```

df['Order Date']=pd.to_datetime(df['Order Date'])
sales=df[['Order Date','Sales']].copy()
sales=sales.groupby('Order Date')['Sales'].sum().reset_index()
sales=sales.set_index('Order Date').sort_index()
sales.head()
```

```

plt.figure(figsize=(12,5))
plt.plot(sales.index,sales['Sales'])
plt.title('Sales Trend')
plt.show()
```

```

monthly=sales.resample('M').sum()
decomp=seasonal_decompose(monthly['Sales'],model='additive',period=12)
decomp.plot();
plt.show()
```

```

plot_acf(monthly['Sales'].dropna()); plt.show()
plot_pacf(monthly['Sales'].dropna()); plt.show()
```

```

train=monthly[:-12]
test=monthly[-12:]
model=ARIMA(train['Sales'],order=(1,1,1))
fit=model.fit()
forecast=fit.forecast(steps=len(test))
```

```

plt.figure(figsize=(12,5))
plt.plot(train.index,train['Sales'],label='Train')
plt.plot(test.index,test['Sales'],label='Actual')
plt.plot(test.index,forecast,label='Forecast')
plt.legend(); plt.show()
```

```

rmse=np.sqrt(mean_squared_error(test['Sales'],forecast))
mae=mean_absolute_error(test['Sales'],forecast)
r2=r2_score(test['Sales'],forecast)
print('RMSE=',rmse)
print('MAE=',mae)
print('R2=',r2)
```

```

residuals=fit.resid
residuals.plot(title='Residuals')
plt.show()
residuals.hist()
plt.show()
```


## Observations
1. Missing values were checked using isnull().sum().
2. Trend visualization shows how sales change over time.
3. Seasonal decomposition identifies trend, seasonal and residual components.
4. ACF/PACF plots justify ARIMA model selection.
5. Forecast values are compared against actual values.
6. RMSE, MAE and R² are used for evaluation.
7. Residual diagnostics verify model adequacy.
