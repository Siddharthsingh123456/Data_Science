```
import random, numpy as np, matplotlib.pyplot as plt

# Q1a Coin Toss
heads=0
trials=10000
for _ in range(trials):
    if random.choice(['H','T'])=='H': heads+=1
print('Heads Probability:',heads/trials)
print('Tails Probability:',1-heads/trials)
```

```
# Q1b Two Dice Sum=7
count=0
trials=10000
for _ in range(trials):
    if random.randint(1,6)+random.randint(1,6)==7:
        count+=1
print('Probability of sum 7:',count/trials)
```

```
# Q2 At least one 6 in 10 rolls
success=0
trials=10000
for _ in range(trials):
    rolls=[random.randint(1,6) for _ in range(10)]
    if 6 in rolls: success+=1
print(success/trials)
```

```
# Q3 Conditional Probability and Bayes
colors=['R']*5+['G']*7+['B']*8
prev=None
blue_prev=0
red_after_blue=0
for _ in range(1000):
    cur=random.choice(colors)
    if prev=='B':
        blue_prev+=1
        if cur=='R': red_after_blue+=1
    prev=cur
print('P(Red|Blue)=',red_after_blue/blue_prev if blue_prev else 0)
```

```
# Q4 Discrete Random Variable
sample=np.random.choice([1,2,3],1000,p=[0.25,0.35,0.4])
print(np.mean(sample),np.var(sample),np.std(sample))
```

```
# Q5 Exponential Distribution
x=np.random.exponential(scale=5,size=2000)
plt.hist(x,bins=30,density=True)
plt.title('Exponential Distribution')
plt.show()
```

```
# Q6 Central Limit Theorem
uniform=np.random.uniform(0,1,10000)
means=[]
for _ in range(1000):
    means.append(np.mean(np.random.choice(uniform,30)))
plt.hist(means,bins=30)
plt.title('Sample Means Distribution')
plt.show()
```
