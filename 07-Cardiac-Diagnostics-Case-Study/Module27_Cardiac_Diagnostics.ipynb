```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```
df = pd.read_csv('dataset_heart.csv')
df.head()
```

```
print("Shape:", df.shape)
print(df.info())
print(df.describe())
print(df.isnull().sum())
```

## Missing Values Analysis

```
plt.figure(figsize=(10,4))
df.isnull().sum().plot(kind='bar')
plt.title('Missing Values by Feature')
plt.show()
```

## Exploratory Data Analysis

```
df['age'].hist(bins=20)
plt.title('Age Distribution')
plt.show()
```

```
sns.heatmap(df.corr(), cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```
df['heart disease'].value_counts().plot(kind='bar')
plt.title('Heart Disease Distribution')
plt.show()
```

## Data Preprocessing

```
X = df.drop('heart disease', axis=1)
y = df['heart disease']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)
```

## Logistic Regression Model

```
model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
```

```
print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6,4))
sns.heatmap(cm, annot=True, fmt='d')
plt.title('Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

```
scores = cross_val_score(model, X_scaled, y, cv=5)

print("Cross Validation Scores:", scores)
print("Mean Accuracy:", scores.mean())
```

## Analysis and Findings

### Dataset Insights
- The dataset contains 270 patient records and multiple cardiac health indicators.
- No missing values were found in the dataset.
- Correlation analysis helps identify features associated with heart disease.

### Model Performance
- Logistic Regression was selected because it is widely used for binary medical classification problems.
- Standardization was applied because features are measured on different scales.
- Accuracy, Precision, Recall and F1-score provide a balanced evaluation.

### Confusion Matrix Interpretation
- True Positives represent correctly identified heart disease cases.
- False Negatives are dangerous because diseased patients may be classified as healthy.
- Recall is important in healthcare because missing a disease can have severe consequences.

### Limitations
- Dataset size is relatively small.
- More advanced models such as Random Forest, XGBoost, or Ensemble Learning may improve performance.
- Hyperparameter tuning could further improve results.

### Conclusion
The model demonstrates that patient attributes such as age, blood pressure, cholesterol, heart rate and ECG-related indicators can be used to predict cardiac disease. Machine learning can assist healthcare professionals in early diagnosis and risk assessment.
